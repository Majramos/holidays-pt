```python
import pandas as pd
from holidays.holidays import Holidays
```

```python
test_holidays = Holidays('2000-01-01', '2100-12-31')
```

```python
(
    pd.DataFrame(test_holidays.calculate().as_list())
        .drop(columns='extent')
        .sort_values('date')
        .assign(
            year=lambda x: x['date'].dt.year,
            month=lambda x: x['date'].dt.month,
            day=lambda x: x['date'].dt.day,
            dayofweek=lambda x: x['date'].dt.dayofweek,
            dayname=lambda x: x['date'].dt.day_name(),
            date=lambda x: x['date'].dt.strftime('%Y%m%d')
        )
        .astype({'label': 'string', 'calculate': 'string', 'date': 'int32',
                 'year': 'int32', 'month': 'int32', 'day': 'int32',
                 'dayofweek': 'int32', 'dayname': 'string'})
# ).to_csv('national_holidays_portugal_2000_2100.csv', index=False)
```
